```python
import pandas as pd

amazon_df = pd.read_csv("amazon.csv", encoding="latin-1")

amazon_df = amazon_df[
    ["Product Name", "Category", "Selling Price", "Image"]
]

amazon_df.head()
```

```python
amazon_df.isnull().sum()
```

```python
amazon_df = amazon_df.dropna()
amazon_df.shape
```

```python
amazon_df["tags"] = (
    amazon_df["Product Name"].astype(str) + " " +
    amazon_df["Category"].astype(str)
)

amazon_df.head()
```

```python
amazon_df = amazon_df[[
    "Product Name",
    "Category",
    "Selling Price",
    "Image",
    "Brand Name",
    "Product Url"
]]
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(
    max_features=5000,
    stop_words="english"
)

vectors = cv.fit_transform(amazon_df["tags"]).toarray()

print(vectors.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vectors)

print(similarity.shape)
```

```python
def recommend(product_name):
    index = amazon_df[amazon_df["Product Name"] == product_name].index[0]

    distances = sorted(
        list(enumerate(similarity[index])),
        reverse=True,
        key=lambda x: x[1]
    )

    for i in distances[1:6]:
        print("--------------------------------")
        print("Product :", amazon_df.iloc[i[0]]["Product Name"])
        print("Category:", amazon_df.iloc[i[0]]["Category"])
        print("Price   :", amazon_df.iloc[i[0]]["Selling Price"])
```

```python
recommend(amazon_df.iloc[0]["Product Name"])
```

```python
import pickle

pickle.dump(amazon_df, open("products.pkl", "wb"))
pickle.dump(similarity, open("similarity.pkl", "wb"))

print("Files Saved Successfully!")
```

```python
import pandas as pd

# Load Dataset
full_df = pd.read_csv("amazon.csv", encoding="latin-1")

# Dataset Information
print("="*60)
print("DATASET SHAPE")
print("="*60)
print(full_df.shape)

print("\n")

print("="*60)
print("COLUMN NAMES")
print("="*60)
print(full_df.columns.tolist())

print("\n")

# Select Required Columns
amazon_df = full_df[
    [
        "Product Name",
        "Brand Name",
        "Category",
        "Selling Price",
        "About Product",
        "Image",
        "Product Url"
    ]
].copy()

# Fill Missing Values
amazon_df["Product Name"] = amazon_df["Product Name"].fillna("")
amazon_df["Brand Name"] = amazon_df["Brand Name"].fillna("")
amazon_df["Category"] = amazon_df["Category"].fillna("")
amazon_df["About Product"] = amazon_df["About Product"].fillna("")
amazon_df["Image"] = amazon_df["Image"].fillna("")
amazon_df["Product Url"] = amazon_df["Product Url"].fillna("")
amazon_df["Selling Price"] = amazon_df["Selling Price"].fillna("")

# Create Tags
amazon_df["tags"] = (
    amazon_df["Product Name"] + " " +
    amazon_df["Brand Name"] + " " +
    amazon_df["Category"] + " " +
    amazon_df["About Product"]
)

print("="*60)
print("FINAL DATAFRAME")
print("="*60)

print("Shape :", amazon_df.shape)

print("\nColumns :")
print(amazon_df.columns.tolist())

print("\nFirst 5 Rows :")
print(amazon_df.head())

print("\n")

print("="*60)
print("MISSING VALUES")
print("="*60)

print(amazon_df.isnull().sum())
```

```python
import streamlit as st
import pandas as pd
import pickle
from pathlib import Path

st.set_page_config(
    page_title="Amazon Product Recommendation System",
    page_icon="🛒",
    layout="wide"
)

st.title("🛒 Amazon Product Recommendation System")
st.write("Find similar Amazon products using Machine Learning.")
st.divider()

PRODUCT_FILE = "products.pkl"
SIMILARITY_FILE = "similarity.pkl"

if not Path(PRODUCT_FILE).exists():
    st.error("products.pkl file not found.")
    st.stop()

if not Path(SIMILARITY_FILE).exists():
    st.error("similarity.pkl file not found.")
    st.stop()

try:
    with open(PRODUCT_FILE, "rb") as f:
        products = pickle.load(f)

    with open(SIMILARITY_FILE, "rb") as f:
        similarity = pickle.load(f)

except Exception as e:
    st.error(f"Unable to load pickle files.\n\n{e}")
    st.stop()

required_columns = [
    "Product Name",
    "Brand Name",
    "Category",
    "Selling Price",
    "About Product",
    "Image",
    "Product Url"
]

missing = [c for c in required_columns if c not in products.columns]

if missing:
    st.error(f"Missing Columns:\n{missing}")
    st.stop()

st.sidebar.header("Filter Products")

categories = sorted(products["Category"].dropna().unique())

selected_category = st.sidebar.selectbox(
    "Select Category",
    ["All"] + list(categories)
)

if selected_category == "All":
    filtered_products = products.copy()
else:
    filtered_products = products[
        products["Category"] == selected_category
    ]

if filtered_products.empty:
    st.warning("No products found.")
    st.stop()

selected_product = st.selectbox(
    "🔍 Search Product",
    filtered_products["Product Name"].tolist()
)

recommend_btn = st.button("🚀 Recommend Products")


def recommend_products(product_name, n=5):

    idx = products.index[
        products["Product Name"] == product_name
    ]

    if len(idx) == 0:
        return pd.DataFrame()

    index = idx[0]

    scores = list(enumerate(similarity[index]))

    scores = sorted(
        scores,
        key=lambda x: x[1],
        reverse=True
    )

    recommendation_index = []

    for i, score in scores:

        if i == index:
            continue

        recommendation_index.append(i)

        if len(recommendation_index) == n:
            break

    return products.iloc[recommendation_index]


def show_image(url, width=200):

    try:
        if pd.notna(url) and str(url).startswith("http"):
            st.image(url, width=width)
        else:
            st.info("No Image Available")
    except:
        st.info("Image cannot be displayed")


if recommend_btn:

    product = products[
        products["Product Name"] == selected_product
    ]

    if product.empty:
        st.error("Product not found.")
        st.stop()

    product = product.iloc[0]

    st.subheader("Selected Product")

    col1, col2 = st.columns([1, 2])

    with col1:
        show_image(product["Image"], 250)

    with col2:

        st.markdown(f"### {product['Product Name']}")
        st.write("**Brand:**", product["Brand Name"])
        st.write("**Category:**", product["Category"])
        st.write("**Price:**", product["Selling Price"])
        st.write(product["About Product"])

        if pd.notna(product["Product Url"]):
            st.link_button("🛒 Buy on Amazon", product["Product Url"])

    st.divider()

    st.subheader("Recommended Products")

    recommendations = recommend_products(selected_product)

    if recommendations.empty:
        st.warning("No recommendations found.")

    else:

        for _, item in recommendations.iterrows():

            c1, c2 = st.columns([1, 3])

            with c1:
                show_image(item["Image"], 150)

            with c2:

                st.markdown(f"#### {item['Product Name']}")
                st.write("**Brand:**", item["Brand Name"])
                st.write("**Category:**", item["Category"])
                st.write("**Price:**", item["Selling Price"])
                st.write(item["About Product"])

                if pd.notna(item["Product Url"]):
                    st.link_button(
                        "🛒 Buy on Amazon",
                        item["Product Url"],
                        key=f"buy_{_}"
                    )

            st.divider()
```
